# A. POSITIONAL AND LABEL-BASED SLICING

**Complete list and shape of the name of all cars**

```python
cars = pd.read_csv('cars.csv')
cars.loc[0:100,['Model']]
```

```python
display(cars.loc[0:100,['Model']].shape)
```

```python
cars_6_to_10 = cars.iloc[5:10,[0,1,2,4,10]]
```

**cars_6_to_10  with Model, mpg, cyl, hp, and gear**

```python
display(cars_6_to_10)
```

# B. MODEL LOOKUP

```python
Toyota_Corolla = cars.loc[cars['Model']=='Toyota Corolla']
Pontiac_Firebird = cars.loc[(cars['Model']=='Pontiac Firebird'),['Model','mpg','hp','wt']]
```

**Toyota Corolla**

```python
display(Toyota_Corolla)
```

**Pontiac Firebird**

```python
display(Pontiac_Firebird)
```

# C. MULTI-MODEL SUBSETTING

```python
selected_cars = pd.DataFrame(cars.loc[
    (cars['Model'] == 'Datsun 710') |
    (cars['Model'] == 'Lotus Europa') |
    (cars['Model'] == 'Ferrari Dino'),
    ['Model', 'mpg', 'cyl', 'hp', 'gear']])
```

**Datsun 710, Lotus Europa, Ferrari Dino with Model, mpg, cyl, hp, and gear with the shape**

```python
display(selected_cars)
```
